## 什麼是提示詞工程？

**提示詞工程**是為像 Claude 這樣的大型語言模型，撰寫有效指示、以產生想要輸出的**藝術與科學**。提示詞工程的核心，是設計、修改和優化給模型的文字輸入（提示詞），以引出準確、相關、有用的回答。它的重點是用一種能讓模型在特定任務上的**理解和表現最大化**的方式，和 Claude「溝通」。提示詞工程的重要性怎麼強調都不為過：

* **提升 AI 的能力**：設計良好的提示詞，能大幅提升 AI 的表現，讓它能更準確、更有效率地處理複雜的任務。
* **連接人類意圖和 AI 輸出之間的落差**：提示詞工程幫助把人類的目標，翻譯成 AI 模型能有效理解並執行的語言。
* **優化資源使用**：熟練的提示詞工程可以減少 token 用量，在正式上線的環境中降低成本、縮短回應時間。

### 它和「基本提示」有什麼不同？

我們把「基本提示」定義為：單純問 AI 模型一個問題，或給它一個簡單的指示。相對地，提示詞工程是一個更精細、更刻意的過程。

* **複雜度**：基本提示通常是用簡單問題進行的單回合互動。提示詞工程則可能包含多回合對話、複雜的指示，以及精心組織的輸入和輸出。
* **精確度**：基本提示可能很模糊、有歧義，導致結果不一致。這在一次性的情況下可能沒關係，但**沒辦法擴大到正式上線的使用情境**。經過工程設計的提示詞很精確，幾乎不留給模型誤解的空間。
* **反覆修正**：基本提示可能是一次性的活動，提示詞工程則包含**有系統地測試、分析，並隨著時間持續改進**提示詞。
* **可擴展性**：提示詞工程的目標，是打造能處理各式各樣輸入和使用情境的提示詞，讓它們適合用在正式上線的環境。

本質上，提示詞工程把和模型的互動，從隨意的聊天，提升為一場精心安排的交流，目的是讓模型在解決真實世界問題時**可以穩定重複地**發揮最大潛力。隨著課程進行，你會學到掌握這項關鍵技能所需的技巧和心態。

> 🧩 **白話小補充**：用做菜來比喻——「基本提示」像是隨手煮一碗泡麵，好不好吃看運氣；「提示詞工程」像是餐廳研發一道招牌菜：要寫食譜、反覆試做、請人試吃、調整配方，確保**每一天、每一位廚師**做出來的味道都一樣好。

## 提示詞工程的生命週期

如果能坐在一張白紙前，第一次就寫出完美的提示詞當然很好，但現實是：**提示詞工程是一個反覆循環的過程**，包含建立、測試和修正提示詞，以達到最佳的表現。

了解這個生命週期，對於寫出有效的提示詞、以及排除出現的問題非常關鍵。

1. 建立初始提示詞
2. 測試並找出問題
3. 選擇適當的技巧
4. 實作改進
5. 反覆修正

![提示詞工程的生命週期圖](../../real_world_prompting/images/prompt_eng_lifecycle.png)

我們來詳細談談每一個部分。

### 1. 建立初始提示詞

先寫一個清楚表達最終目標的「**初稿**」提示詞。理想上，這第一次嘗試會用上一些我們之前介紹過的提示詞技巧：定義目標、找出關鍵資訊、好好組織提示詞等等。不管怎樣，最終版本的提示詞都**不太可能**和這份初稿長得一樣。

### 2. 測試並找出問題

有了基本的提示詞之後，下一步是用**各種可能的輸入**來測試它。我們接下來的「提示詞評估」課程（本套課程的第 4 單元）會詳細介紹這個過程，但重點可以歸納為：
* **準備測試案例**：建立一組多樣化的輸入，涵蓋各種情境和邊緣情況。
* **執行初步測試**：用準備好的輸入執行你的提示詞，觀察輸出。
* **分析結果**：替模型的回答「打分數」。可以用很多種方式：用程式碼、靠人類的專業判斷，或用大型語言模型來評分。

在整個測試過程中，要記住的關鍵指標包括：
* **準確度**：輸出正確、相關嗎？
* **一致性**：模型在不同的輸入上表現是否類似？它怎麼處理邊緣情況？
* **完整性**：輸出是否包含所有需要的資訊？
* **遵守指示**：模型有沒有遵守所有給定的指示？

> 🧩 **白話小補充：什麼是「邊緣情況（edge case）」？** 就是「不常見、比較極端或奇怪的輸入」。例如請 Claude 摘要病歷，一般病歷很正常，但如果病歷是空的？只有一行？用別的語言寫？有錯字？這些就是邊緣情況，最容易讓提示詞出錯。

### 3. 選擇適當的技巧

測試完初始提示詞、找出具體的問題之後，就是修改提示詞的時候了。與其**盲目地**做改進，理想的做法包含下面這些步驟：

* **診斷根本原因**：針對每一個找到的問題，試著了解它**為什麼**會發生。
* **研究解決方法**：根據你的診斷，探索可能解決問題的提示詞工程技巧。
* **選擇技巧**：挑出最有希望的策略來實作。

### 4. 實作改進

接著，真的把改進實作到你原本的提示詞中。修改原本的提示詞，加入你選擇的技巧，針對先前找到的問題下手。**如果你要做好幾個修改，可以考慮一次只實作一個**，這樣比較能了解每個修改各自的影響。

> 💡 **小提醒**：這就像做實驗要「控制變因」。如果你一次改了三個地方，結果變好了，你不會知道是哪一個修改有效；結果變差了，也不知道是哪一個害的。

### 5. 反覆修正

重複上面的過程！

* **重新測試**：用最一開始的那些測試案例，執行更新後的提示詞。
* **比較結果**：分析輸出有什麼改變，以及原本針對的問題是否已經解決。
* **找出新問題**：留意你的修改是否帶來了任何新的問題。
* **重複循環**：持續這個「測試、分析、修正」的過程，直到達到令人滿意的表現。

## 📌 重點整理

- **提示詞工程** ≠ 隨便問問：它講究**精確、可重複、可擴展**，並且**持續測試和改進**。
- **生命週期**：① 寫初稿 → ② 用多樣的測試案例測試、打分數 → ③ 診斷原因、選擇技巧 → ④ 實作（一次改一個）→ ⑤ 重新測試、比較、再循環。
- 測試時看四個指標：**準確度、一致性、完整性、遵守指示**。
