## 準備工作

執行下面這個準備格，載入你的 API 金鑰，並建立 `get_completion` 輔助函式。（每一章開頭都有這一格，內容跟[第 1 章](01_Basic_Prompt_Structure.md#準備工作)幾乎一樣。）

```python
%pip install "anthropic<1"

# Import python's built-in regular expression library
import re
import anthropic

# Retrieve the API_KEY & MODEL_NAME variables from the IPython store
%store -r API_KEY
%store -r MODEL_NAME

client = anthropic.Anthropic(api_key=API_KEY)

# Note that we changed max_tokens to 4K just for this lesson to allow for longer completions in the exercises
def get_completion(prompt: str, system_prompt=""):
    message = client.messages.create(
        model=MODEL_NAME,
        max_tokens=4000,
        temperature=0.0,
        system=system_prompt,
        messages=[
          {"role": "user", "content": prompt}
        ]
    )
    return message.content[0].text
```

（這一章特別把 `max_tokens` 調高到 4000，讓練習可以產生比較長的回答。程式碼裡的註解就是在說這件事。）

## 課程

**Claude 對「清楚、直接」的指示反應最好。**

把 Claude 想成一個**剛到職的新同事**。除了你明確告訴它的事情以外，**Claude 對要做什麼完全沒有任何背景資訊**。就像你第一次教一個人做某項工作一樣，你越是**直截了當地**解釋你到底想要什麼，Claude 的回答就會越好、越準確。

拿不定主意的時候，就遵守**清楚提示詞的黃金法則**：
- 把你的提示詞拿給一位同事或朋友看，請他們**自己照著指示做做看**，看他們能不能做出你想要的結果。**如果他們看得一頭霧水，Claude 也會一頭霧水。**

> 🧩 **白話小補充**：很多人寫提示詞時，會以為 Claude「應該懂我的意思」。但 Claude 只看得到你打的字，看不到你腦中的想法。所以要把「你覺得理所當然」的要求也寫出來：要多長？要什麼格式？要給誰看？不要什麼？

### 範例

我們拿「寫詩」這種任務來舉例。（請忽略音節數不符的地方——LLM 目前還不太會數音節。）

```python
# Prompt
PROMPT = "Write a haiku about robots."

# Print Claude's response
print(get_completion(PROMPT))
```

（「寫一首關於機器人的俳句。」）

這首俳句還不錯，但使用者可能希望 Claude **直接開始寫詩**，不要有「這是一首俳句：」之類的開場白。

要怎麼做到呢？我們就**直接要求**！

```python
# Prompt
PROMPT = "Write a haiku about robots. Skip the preamble; go straight into the poem."

# Print Claude's response
print(get_completion(PROMPT))
```

（「寫一首關於機器人的俳句。**跳過開場白，直接開始寫詩。**」）

再看另一個例子。我們問 Claude：史上最偉大的籃球員是誰？你會看到，Claude 雖然列出了幾個名字，但**沒有給出一個明確的「最佳」答案**。

```python
# Prompt
PROMPT = "Who is the best basketball player of all time?"

# Print Claude's response
print(get_completion(PROMPT))
```

（「史上最偉大的籃球員是誰？」）

我們能讓 Claude 下定決心、選出一位最佳球員嗎？可以！**直接問就好！**

```python
# Prompt
PROMPT = "Who is the best basketball player of all time? Yes, there are differing opinions, but if you absolutely had to pick one player, who would it be?"

# Print Claude's response
print(get_completion(PROMPT))
```

（「史上最偉大的籃球員是誰？**對，大家的意見不同，但如果你一定要選一位球員，你會選誰？**」）

### 練習 2.1 - 西班牙文

修改 `SYSTEM_PROMPT`，讓 Claude 用**西班牙文**輸出答案。

```python
# System prompt - this is the only field you should change
SYSTEM_PROMPT = "[Replace this text]"

# Prompt
PROMPT = "Hello Claude, how are you?"

# Get Claude's response
response = get_completion(PROMPT, SYSTEM_PROMPT)

# Function to grade exercise correctness
def grade_exercise(text):
    return "hola" in text.lower()

# Print Claude's response and the corresponding grade
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```

**批改程式在做什麼？** 檢查回答（轉成小寫後）裡有沒有 `hola`（西班牙文的「你好」）。
（使用者的問題是：「哈囉 Claude，你好嗎？」）

### 練習 2.2 - 只要一位球員

修改 `PROMPT`，讓 Claude **完全不模稜兩可**，並且**只**回答一位特定球員的名字，**不要有任何其他的字或標點符號**。

```python
# Prompt - this is the only field you should change
PROMPT = "[Replace this text]"

# Get Claude's response
response = get_completion(PROMPT)

# Function to grade exercise correctness
def grade_exercise(text):
    return text == "Michael Jordan"

# Print Claude's response and the corresponding grade
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```

**批改程式在做什麼？** `text == "Michael Jordan"`：回答必須**完全等於** `Michael Jordan`（麥可·喬丹）——多一個字、多一個句點、多一個空白都不行。

> 💡 這題的批改很嚴格，練習的重點就是學會「明確告訴 Claude 輸出格式」。

### 練習 2.3 - 寫一個故事

修改 `PROMPT`，讓 Claude 盡可能寫出**越長越好**的回答。如果你的答案**超過 800 個字**，Claude 的回答就會被判定為正確。

```python
# Prompt - this is the only field you should change
PROMPT = "[Replace this text]"

# Get Claude's response
response = get_completion(PROMPT)

# Function to grade exercise correctness
def grade_exercise(text):
    trimmed = text.strip()
    words = len(trimmed.split())
    return words >= 800

# Print Claude's response and the corresponding grade
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```

**批改程式在做什麼？** `text.strip().split()` 會用空白把文字切成一個一個的英文單字，`len(...)` 算出有幾個單字，至少 800 個才算過關。

> ⚠️ 注意：批改程式是「用空白切開來數」，**中文字之間沒有空白**，所以一整段中文可能只會被算成很少的「字」。這題請讓 Claude **用英文寫故事**。

## 📌 重點整理
